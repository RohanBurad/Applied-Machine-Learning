**Python program to demonstrate basic input/output and arithmetic operations.**

```python
# CHOICE BASED CALCI
print("-------------------------------------")
print("   WELCOME TO CHOICE BASED CALCULATOR")
print("-------------------------------------\n")

# ARITHMETIC OPERATIONS FUNCTION
def add(A,B):
  Sum = A + B
  return Sum
def sub(A,B):
  Sub = A - B
  return Sub
def multi(A,B):
  Mul = A * B
  return Mul
def div(A,B):
  divi = A / B
  return divi

# MENU FUNCTION
def menu_sel(choice):
  match choice:
    case '1':
      print(" CHOICE 1 SELECTED SUCCESSFULLY.\n")
      A = float(input("Enter the value of A : "))
      B = float(input("Enter the value of B : "))
      Sum = add(A,B)
      print(f"The Addition of A & B is {Sum}.\n")
    case '2':
      print(" CHOICE 2 SELECTED SUCCESSFULLY.\n")
      A = float(input("Enter the value of A : "))
      B = float(input("Enter the value of B : "))
      Sub = sub(A,B)
      print(f"The Subtraction of A & B is {Sub}.\n")
    case '3':
      print(" CHOICE 3 SELECTED SUCCESSFULLY.\n")
      A = float(input("Enter the value of A : "))
      B = float(input("Enter the value of B : "))
      Mul = multi(A,B)
      print(f"The Multiple of A & B is {Mul}.\n")
    case '4':
      print(" CHOICE 4 SELECTED SUCCESSFULLY.\n")
      A = float(input("Enter the value of A : "))
      B = float(input("Enter the value of B : "))
      if B==0:
        print("The Division is not possible by zero.\n")
      else:
        Div = div(A,B)
      print(f"The Division of A by B is {Div}.")

#   MENU DISPLAY
while True:
  print("-------------------------------------")
  print("               MENU")
  print("-------------------------------------")
  print("1. Addition.")
  print("2. Subtraction.")
  print("3. Multiply.")
  print("4. Divide.")
  print("5. Quit")
  print("-------------------------------------\n")
  print("-------------------------------------")
  choice = input("Enter the choice : ").strip()
  print("-------------------------------------\n")
  menu_sel(choice)
  if choice == '5':
    print("-------------------------------------")
    print("Quit Successfully.")
    print("-------------------------------------\n")
    break
```
