**Write a python program to do operation on list (Insert , Delete , Search)**

```python
# INSERTION , DELETION , SEARCH OPERATIONS ON A LIST.
# ------------------------------------------------------------------
print("-------------------------------------")
print("     WELCOME TO THE LIST MAKER")
print("-------------------------------------\n")
# ------------------------------------------------------------------
List = []
# METHOD FUNCTIONS
def for_loop():
  if not List:
    print("List is Empty.")
  else:
    print("\n---Your List---\n")
    for item in List:
      print(item)
    print("-------------------\n")
  input("Press Enter to return to the menu...")
 # ------------------------------------------------------------------
def insert():
  item = input("Add the Item (Separate the item by space) : ").strip()
  if not item:
    print("No items entered !")
    return
  List.extend(item.split())
  input("Press Enter to return to the menu...")
# ------------------------------------------------------------------
def delete():
  try:
    B = input("Enter the item to be deleted : ")
    List.remove(B)
    print("Item Deleted")
  except (ValueError,IndexError):
    print(f"Invalid Index!")
  input("Press Enter to return to the menu...")
# ------------------------------------------------------------------
def search():
  item = input("Enter the item you are searching : ")
  try:
    Index = List.index(item)
    print(f"{item} is at {Index} of the List.")
  except ValueError:
    print(f"{item} is not in the List.")
  input("Press Enter to return to the menu...")
# ------------------------------------------------------------------
def count():
  print(f"Total items in the List: {len(List)}.")
  input("Press Enter to return to the menu...")
# ------------------------------------------------------------------
def clear():
  List.clear()
  print("---------- List Successfully Cleared ----------\n")
  input("Press Enter to return to the menu...")
# ------------------------------------------------------------------
# MENU FUNCTION
def menu_selection(opera):
  match opera:
    case '1':
      insert()
    case '2':
      delete()
    case '3':
      search()
    case '4':
      count()
    case '5':
      clear()
    case '6':
      for_loop()
    case '7':
      print(" THANKS FOR USING OUR SERVICE ")
      return False
    case _:
      print("Invalid Option.")
  return True
# MENU
while True:
  print("-------------------------------------")
  print("                MENU")
  print("-------------------------------------")
  print("1. ADD ITEM.")
  print("2. DELETE ITEM.")
  print("3. SEARCH ITEM.")
  print("4. LIST COUNT.")
  print("5. CLEAR THE LIST.")
  print("6. DISPLAY THE LIST")
  print("7. QUIT THE LIST MAKER.")
  print("-------------------------------------\n")
  print("-------------------------------------")
  opera = input("Enter the choice Number : ")
  print("-------------------------------------\n")
  if not menu_selection(opera):
    break
```
